### Section 1: Core Concepts & Applications

**1. Which industries commonly use Deep Learning for image recognition applications?**
- **Answer:** **d. All of the above**
- **Explanation:** Image recognition is widely used in Healthcare (MRI/X-ray), Manufacturing (defect detection), and Retail (inventory tracking).

**2. What distinguishes Deep Learning from traditional Machine Learning?**
- **Answer:** **i. a,d**
- **Explanation:** Deep learning involves greater neural network depth (a) and requires significantly larger volumes of data (d) compared to traditional ML.

**3. Which metrics are commonly used to evaluate the performance of a classification model?**
- **Answer:** **c. F1 score**
- **Explanation:** While MSE and RMSE are for regression, F1 score is a key metric for classification tasks.

**4. Which of the following are the key components of Deep Learning (DL)?**
- **Answer:** **b. Neural Networks**
- **Explanation:** Neural networks are the foundational architecture of all deep learning algorithms.

**5. What are the main challenges in training deep neural networks?**
- **Answer:** **b. Overfitting**
- **Explanation:** Deep networks are prone to capturing noise in the training data, leading to poor generalization.

### Section 2: Project Lifecycle & Monitoring

**6. Which technology is a subset of AI and focuses on training models to learn from data?**
- **Answer:** **b. Machine Learning**
- **Explanation:** Machine Learning is the parent field of Deep Learning and a subset of AI.

**7. In the context of Deep Learning, what is a neural network?**
- **Answer:** **b. A network of interconnected neurons**
- **Explanation:** It is inspired by the biological structure of the human brain.

**8. What is the role of hyperparameter tuning in the model evaluation phase?**
- **Answer:** **b. To optimize model parameters for better performance**
- **Explanation:** Tuning ensures the model reaches its peak potential by adjusting settings like learning rate.

**9. What is the primary goal during the project initiation phase of a Deep Learning project?**
- **Answer:** **b. Define the problem and scope**
- **Explanation:** Every project must start with a clear objective and boundaries.

**10. Why is continuous monitoring important in the life cycle of a deployed Deep Learning model?**
- **Answer:** **b. To ensure the model adapts to changing data patterns**
- **Explanation:** Real-world data drifts over time, requiring models to be monitored for accuracy drops.
