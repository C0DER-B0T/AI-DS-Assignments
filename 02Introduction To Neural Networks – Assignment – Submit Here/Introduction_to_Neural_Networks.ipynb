<h1>Module 2: Introduction to Neural Networks - Assignment</h1>
<h3>Objective: Predict y = 3x1 + 4x2 using Simple Neural Network</h3>
<hr>

### Step 1: Imports and Data Initialization

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample Data
x1 = 3
x2 = 2
y_actual = 17

print(f"Target: y = 3*{x1} + 4*{x2} = {y_actual}")
```

### Step 2: Weight Initialization and Hyperparameters

```python
# Initialize random weights between 1 and 10
np.random.seed(42)
w1 = np.random.uniform(1, 10)
w2 = np.random.uniform(1, 10)

learning_rate = 0.01
epochs = 20

print(f"Initial Weights: w1={w1:.4f}, w2={w2:.4f}")
```

### Step 3: Training Loop (Feed Forward & Back Propagation)

```python
errors = []
y_preds = []

for epoch in range(epochs):
    # --- Feed Forward ---
    y_pred = x1 * w1 + x2 * w2
    y_preds.append(y_pred)
    
    # --- Error Calculation (MSE) ---
    error = (y_actual - y_pred)**2
    errors.append(error)
    
    # --- Back Propagation (Gradient Descent) ---
    # Gradient of Error w.r.t weights
    dw1 = -2 * (y_actual - y_pred) * x1
    dw2 = -2 * (y_actual - y_pred) * x2
    
    # Update Weights
    w1 = w1 - learning_rate * dw1
    w2 = w2 - learning_rate * dw2
    
    if (epoch + 1) % 5 == 0:
        print(f"Epoch {epoch+1:02d}: Error = {error:.6f}, y_pred = {y_pred:.4f}")
```

### Step 4: Final Results and Convergence

```python
print(f"Final Predicted y: {y_pred:.4f} (Actual: {y_actual})")
print(f"Final Weights: w1={w1:.4f}, w2={w2:.4f}")
print("Note: Ideally w1 -> 3.0 and w2 -> 4.0 if trained longer.")
```

### Step 5: Visualization of Learning Process

```python
plt.figure(figsize=(14, 6))

# Plot Error
plt.subplot(1, 2, 1)
plt.plot(range(1, epochs + 1), errors, marker='o', color='crimson', linestyle='--')
plt.title('Error Reduction over Epochs', fontsize=14)
plt.xlabel('Epochs')
plt.ylabel('Squared Error')
plt.grid(True, alpha=0.3)

# Plot Predictions
plt.subplot(1, 2, 2)
plt.plot(range(1, epochs + 1), y_preds, marker='s', color='royalblue', label='Predicted Y')
plt.axhline(y=y_actual, color='forestgreen', linestyle='-', label=f'Actual Y ({y_actual})')
plt.title('Prediction Convergence', fontsize=14)
plt.xlabel('Epochs')
plt.ylabel('y_pred')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
