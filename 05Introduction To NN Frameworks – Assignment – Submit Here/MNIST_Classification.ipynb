### Step 1: Imports and Data Preprocessing

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize image pixel values (0-255 -> 0-1)
X_train, X_test = X_train / 255.0, X_test / 255.0

# Convert labels to one-hot encoding
y_train_cat = to_categorical(y_train)
y_test_cat = to_categorical(y_test)

print(f"Training images: {X_train.shape}")
print(f"Test images: {X_test.shape}")
```

### Step 2: Model Development

```python
model = Sequential([
    Flatten(input_shape=(28, 28)), # Flatten 2D image to 1D vector
    Dense(128, activation='relu'), # Hidden Layer 1
    Dense(64, activation='relu'),  # Hidden Layer 2
    Dense(10, activation='softmax') # Output Layer (10 digits)
])

model.compile(optimizer='sgd',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

### Step 4: Prediction Function

```python
import numpy as np

def predict_digit(image_index):
    """Predict the digit of an image from the test set"""
    img = X_test[image_index]
    prediction = model.predict(img.reshape(1, 28, 28))
    predicted_label = np.argmax(prediction)
    actual_label = y_test[image_index]
    
    print(f"Predicted: {predicted_label}, Actual: {actual_label}")
    return predicted_label
```
