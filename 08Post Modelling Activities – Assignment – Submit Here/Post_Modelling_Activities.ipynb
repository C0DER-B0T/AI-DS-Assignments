### Step 1: Imports and Data Preprocessing

```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import TensorBoard
import datetime
import os

# Load and Normalize
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0
y_train_cat = to_categorical(y_train)
y_test_cat = to_categorical(y_test)
```

### Step 2: Model Building and TensorBoard Setup

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(512, activation='relu'),
    Dense(256, activation='relu'),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam', 
              loss='categorical_crossentropy', 
              metrics=['accuracy'])

# TensorBoard Setup
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)

print(f"TensorBoard logs will be saved to: {log_dir}")
```

### Step 3: Model Training and Saving

```python
print("Ready for training and saving...")
model.fit(X_train, y_train_cat, epochs=5, validation_data=(X_test, y_test_cat), callbacks=[tensorboard_callback])

# Save the model
model.save('fashion_mnist_model.h5')
print("Model saved as fashion_mnist_model.h5")
```

### Step 4: Model Loading and Verification

```python
def verify_saved_model(path='fashion_mnist_model.h5'):
    """Load and check the saved model architecture"""
    if os.path.exists(path):
        loaded = load_model(path)
        loaded.summary()
        print("Model verified successfully.")
    else:
        print("Model file not found. Please train and save first.")
```
