<h1>Module 7: Learning from Pre-Built Models - Assignment</h1>
<h3>Task: Cats vs Dogs Classification using Transfer Learning (VGG16)</h3>
<hr>

### Step 1: Imports and VGG16 Base Loading

```python
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import SGD

# Load VGG16 without top layers
vgg_base = VGG16(weights='imagenet', include_top=False, input_shape=(150, 150, 3))

# Freeze the base layers to retain pre-trained weights
vgg_base.trainable = False

print("VGG16 base loaded and frozen.")
```

### Step 2: Custom Classifier Head

```python
model = Sequential([
    vgg_base,
    Flatten(),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(2, activation='softmax') # Cats vs Dogs
])

model.compile(optimizer=SGD(learning_rate=0.001, momentum=0.9),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

### Step 3: Data Augmentation and Training Plan

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    horizontal_flip=True
)

print("ImageDataGenerator configured.")
```

### Step 4: Transfer Learning Prediction Function

```python
import numpy as np
from tensorflow.keras.preprocessing import image

def predict_transfer(img_path):
    """Predict category using the Transfer Learning model"""
    try:
        img = image.load_img(img_path, target_size=(150, 150))
        img_array = image.img_to_array(img) / 255.0
        img_array = np.expand_dims(img_array, axis=0)
        
        prediction = model.predict(img_array)
        class_idx = np.argmax(prediction[0])
        label = 'Dog' if class_idx == 1 else 'Cat'
        confidence = prediction[0][class_idx]
        
        return label, confidence
    except Exception as e:
        return f"Error: {str(e)}"
```
