<h1>Module 8: Recurrent Neural Networks - Assignment</h1>
<h3>Task: IMDB Movie Review Sentiment Analysis using RNN</h3>
<hr>

### Step 1: Imports and Data Loading

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# Load IMDB dataset (top 10,000 words)
num_words = 10000
max_len = 500
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=num_words)

print(f"Loaded {len(X_train)} training sequences and {len(X_test)} test sequences.")
```

### Step 2: Data Preprocessing (Padding)

```python
# Pad sequences to ensure uniform length
X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)

print(f"Padded sequence shape: {X_train.shape}")
```

### Step 3: RNN Model Architecture

```python
model = Sequential([
    Embedding(num_words, 32, input_length=max_len),
    SimpleRNN(32, return_sequences=False),
    Dense(1, activation='sigmoid') # Binary output: Positive (1) or Negative (0)
])

model.compile(optimizer='rmsprop',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.summary()
```

### Step 4: Training and Evaluation Plan

```python

history = model.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.2, verbose=1)


test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {test_acc*100:.2f}%")
```
