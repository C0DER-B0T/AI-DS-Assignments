### Step 1: Loading the COVID-19 Dataset
Loading the clean COVID-19 complete dataset containing global observations.

```python
import pandas as pd

# Extracted Sheet ID from your link
sheet_id = "1f95IiphoNdjXw-jCZbNSEQCdjPorZ7ET0-5T3PLIbIA"
sheet_url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv"

# Read directly into pandas DataFrame
df = pd.read_csv(sheet_url)
df.head()
```

### Step 2: Checking Dataset Structure and Data Types
Inspecting total entries, columns, and data types.

```python
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.info()
```

### Step 3: Checking Missing Values
Checking for null values across all attributes in the dataset.

```python
df.isnull().sum()
```

### Step 4: Date Parsing and Preprocessing
Converting the Date column to standard datetime format and sorting chronologically.

```python
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)
print(f"Observation timeline: {df['Date'].min().strftime('%Y-%m-%d')} to {df['Date'].max().strftime('%Y-%m-%d')}")
```

### Step 5: Aggregating Global Daily Cases
Aggregating cumulative confirmed, death, and recovered cases across all countries by date.

```python
global_daily = df.groupby('Date')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
global_daily.tail()
```

### Step 6: Interactive Global Cumulative Trends with Plotly
Plotting global cumulative confirmed cases, recoveries, active cases, and deaths using Plotly graph objects.

```python
import plotly.graph_objects as go

fig = go.Figure()
fig.add_trace(go.Scatter(x=global_daily['Date'], y=global_daily['Confirmed'], mode='lines', name='Confirmed', line=dict(color='royalblue', width=2)))
fig.add_trace(go.Scatter(x=global_daily['Date'], y=global_daily['Recovered'], mode='lines', name='Recovered', line=dict(color='forestgreen', width=2)))
fig.add_trace(go.Scatter(x=global_daily['Date'], y=global_daily['Active'], mode='lines', name='Active', line=dict(color='darkorange', width=2)))
fig.add_trace(go.Scatter(x=global_daily['Date'], y=global_daily['Deaths'], mode='lines', name='Deaths', line=dict(color='crimson', width=2)))

fig.update_layout(
    title='Worldwide COVID-19 Cumulative Cases Progression',
    xaxis_title='Date',
    yaxis_title='Number of Cases',
    template='plotly_white',
    hovermode='x unified'
)
fig.show()
```

### Step 7: Calculating Daily New Cases
Computing daily incremental confirmed cases, deaths, and recoveries to observe daily trajectory and infection spikes.

```python
global_daily['Daily_Confirmed'] = global_daily['Confirmed'].diff().fillna(0)
global_daily['Daily_Deaths'] = global_daily['Deaths'].diff().fillna(0)
global_daily['Daily_Recovered'] = global_daily['Recovered'].diff().fillna(0)

# Apply 7-day rolling average for smoother trend visibility
global_daily['Daily_Confirmed_7d_avg'] = global_daily['Daily_Confirmed'].rolling(window=7).mean()

global_daily[['Date', 'Daily_Confirmed', 'Daily_Confirmed_7d_avg']].tail()
```

### Step 8: Visualizing Daily New Confirmed Cases
Plotting daily new confirmed cases along with the 7-day rolling average using Plotly.

```python
import plotly.express as px

fig = go.Figure()
fig.add_trace(go.Bar(x=global_daily['Date'], y=global_daily['Daily_Confirmed'], name='Daily Confirmed', marker_color='lightblue'))
fig.add_trace(go.Scatter(x=global_daily['Date'], y=global_daily['Daily_Confirmed_7d_avg'], mode='lines', name='7-Day Rolling Avg', line=dict(color='red', width=2)))

fig.update_layout(
    title='Daily New Confirmed COVID-19 Cases Worldwide',
    xaxis_title='Date',
    yaxis_title='Daily New Cases',
    template='plotly_white'
)
fig.show()
```

### Step 9: Top 10 Most Impacted Countries
Extracting data from the latest recorded date to identify the top 10 countries by confirmed cases.

```python
latest_date = df['Date'].max()
latest_df = df[df['Date'] == latest_date]

country_summary = latest_df.groupby('Country/Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
top10_confirmed = country_summary.sort_values('Confirmed', ascending=False).head(10)

fig = px.bar(
    top10_confirmed,
    x='Confirmed',
    y='Country/Region',
    orientation='h',
    title=f'Top 10 Most Affected Countries by Confirmed Cases (as of {latest_date.strftime("%Y-%m-%d")})',
    color='Confirmed',
    color_continuous_scale='Reds'
)
fig.update_layout(yaxis=dict(autorange="reversed"), template='plotly_white')
fig.show()
```

### Step 10: Infection Mortality and Recovery Rates
Evaluating the Recovery Rate and Mortality (CFR) across top affected nations to compare healthcare outcome trends.

```python
top10_confirmed['Recovery_Rate_%'] = (top10_confirmed['Recovered'] / top10_confirmed['Confirmed']) * 100
top10_confirmed['Mortality_Rate_%'] = (top10_confirmed['Deaths'] / top10_confirmed['Confirmed']) * 100

top10_confirmed[['Country/Region', 'Confirmed', 'Deaths', 'Recovered', 'Recovery_Rate_%', 'Mortality_Rate_%']].round(2)
```

### Step 11: In-Depth Trend Analysis: India
Filtering records for India to examine national transmission dynamics, active cases, and recovery curves.

```python
india_df = df[df['Country/Region'] == 'India'].groupby('Date')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
india_df['Daily_Confirmed'] = india_df['Confirmed'].diff().fillna(0)
india_df.tail()
```

### Step 12: Visualizing India COVID-19 Trajectory with Plotly
Plotting cumulative confirmed cases, recoveries, and active cases specifically for India.

```python
fig_india = go.Figure()
fig_india.add_trace(go.Scatter(x=india_df['Date'], y=india_df['Confirmed'], mode='lines', name='Confirmed', line=dict(color='orange', width=2)))
fig_india.add_trace(go.Scatter(x=india_df['Date'], y=india_df['Recovered'], mode='lines', name='Recovered', line=dict(color='green', width=2)))
fig_india.add_trace(go.Scatter(x=india_df['Date'], y=india_df['Active'], mode='lines', name='Active', line=dict(color='firebrick', width=2)))

fig_india.update_layout(
    title='COVID-19 Progression in India',
    xaxis_title='Date',
    yaxis_title='Cases',
    template='plotly_white',
    hovermode='x unified'
)
fig_india.show()
```

### Step 13: Preparing Data for Time Series Forecasting with Prophet
Facebook Prophet requires a specific dataframe format:
- `ds`: datestamp column
- `y`: metric to predict (confirmed cases)

```python
prophet_df = global_daily[['Date', 'Confirmed']].rename(columns={'Date': 'ds', 'Confirmed': 'y'})
prophet_df.head()
```

### Step 14: Fitting Prophet Model on Global Confirmed Cases
Initializing and fitting the Prophet forecasting model with 95% prediction intervals.

```python
from prophet import Prophet

model_global = Prophet(interval_width=0.95, daily_seasonality=False, weekly_seasonality=True, yearly_seasonality=False)
model_global.fit(prophet_df)
print("Prophet global model fitted successfully.")
```

### Step 15: Forecasting Global Cases for the Next 7 Days
Creating future dates (7 days ahead) and generating predictions along with upper and lower confidence bounds.

```python
future_global = model_global.make_future_dataframe(periods=7)
forecast_global = model_global.predict(future_global)

forecast_global[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(10)
```

### Step 16: Interactive Visualization of Global Forecast with Plotly
Plotting historical actual cases alongside predicted values and the 95% uncertainty interval.

```python
fig_forecast = go.Figure()

# Actual historical data
fig_forecast.add_trace(go.Scatter(x=prophet_df['ds'], y=prophet_df['y'], mode='markers', name='Actual Cases', marker=dict(color='black', size=4)))

# Predicted trajectory
fig_forecast.add_trace(go.Scatter(x=forecast_global['ds'], y=forecast_global['yhat'], mode='lines', name='Forecasted Cases', line=dict(color='blue', width=2)))

# Upper & lower confidence intervals
fig_forecast.add_trace(go.Scatter(
    x=pd.concat([forecast_global['ds'], forecast_global['ds'][::-1]]),
    y=pd.concat([forecast_global['yhat_upper'], forecast_global['yhat_lower'][::-1]]),
    fill='toself',
    fillcolor='rgba(0,100,80,0.2)',
    line=dict(color='rgba(255,255,255,0)'),
    hoverinfo="skip",
    showlegend=True,
    name='95% Confidence Interval'
))

fig_forecast.update_layout(
    title='Global COVID-19 Confirmed Cases: Prophet 7-Day Prediction',
    xaxis_title='Date',
    yaxis_title='Cumulative Confirmed Cases',
    template='plotly_white'
)
fig_forecast.show()
```

### Step 17: Forecasting COVID-19 Confirmed Cases for India (Next 7 Days)
Preparing India's time series data, fitting Prophet, and making 7-day ahead forecasts.

```python
prophet_india_df = india_df[['Date', 'Confirmed']].rename(columns={'Date': 'ds', 'Confirmed': 'y'})

model_india = Prophet(interval_width=0.95, daily_seasonality=False, weekly_seasonality=True, yearly_seasonality=False)
model_india.fit(prophet_india_df)

future_india = model_india.make_future_dataframe(periods=7)
forecast_india = model_india.predict(future_india)

forecast_india[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(7)
```

### Step 18: Visualizing India 7-Day Forecast with Plotly
Comparing observed historical progression in India with the forecasted 7-day trajectory.

```python
fig_india_pred = go.Figure()

fig_india_pred.add_trace(go.Scatter(x=prophet_india_df['ds'], y=prophet_india_df['y'], mode='markers', name='Actual India Cases', marker=dict(color='darkred', size=4)))
fig_india_pred.add_trace(go.Scatter(x=forecast_india['ds'], y=forecast_india['yhat'], mode='lines', name='Forecasted Trend', line=dict(color='darkorange', width=2)))

fig_india_pred.add_trace(go.Scatter(
    x=pd.concat([forecast_india['ds'], forecast_india['ds'][::-1]]),
    y=pd.concat([forecast_india['yhat_upper'], forecast_india['yhat_lower'][::-1]]),
    fill='toself',
    fillcolor='rgba(255,165,0,0.2)',
    line=dict(color='rgba(255,255,255,0)'),
    name='95% Confidence Interval'
))

fig_india_pred.update_layout(
    title='India COVID-19 Confirmed Cases: Prophet 7-Day Ahead Forecast',
    xaxis_title='Date',
    yaxis_title='Confirmed Cases',
    template='plotly_white'
)
fig_india_pred.show()
```

### Step 19: Key Inferences and Summary
- **Transmission Dynamics:** The epidemic demonstrated an exponential growth trajectory across primary global epicenters before stabilizing in regions with strict intervention protocols.
- **Outcome Metrics:** Recovery rates steadily improved over the timeline as healthcare protocols and resource management adapted.
- **Forecasting Performance:** Prophet accurately captures weekly seasonality and current momentum, providing a realistic 7-day projection bounded by 95% confidence intervals.
